# Code Evaluation for Circuit Analysis

This notebook evaluates the code implementation in `/net/scratch2/smallyan/filter_eval`.

## Step 1: Read the Plan and Codewalk Files

```python
# List all the code cells in demo.ipynb and understand the structure
import json
import os

# Read the demo.ipynb
with open('/net/scratch2/smallyan/filter_eval/demo.ipynb', 'r') as f:
    notebook = json.load(f)

# Count and list code cells
code_cells = []
for i, cell in enumerate(notebook['cells']):
    if cell['cell_type'] == 'code':
        source = ''.join(cell['source'])
        # Get first few lines
        preview = source[:200].replace('\n', '\\n')
        code_cells.append({
            'index': i,
            'preview': preview,
            'full_source': source,
            'has_output': len(cell.get('outputs', [])) > 0
        })

print(f"Total cells: {len(notebook['cells'])}")
print(f"Code cells: {len(code_cells)}")
print("\nCode cells preview:")
for j, cell in enumerate(code_cells):
    print(f"\n[Cell {j}] (notebook index {cell['index']}):")
    print(f"  Preview: {cell['preview'][:100]}...")
    print(f"  Has output: {cell['has_output']}")
```
