```python
"""
CS1 EVALUATION: Conclusion vs Original Results

Let's compare the key claims in the documentation with the actual data recorded in the implementation.
"""

print("=" * 80)
print("CS1: CONCLUSION VS ORIGINAL RESULTS EVALUATION")
print("=" * 80)

# 1. AIE values for filter heads
print("\n1. FILTER HEAD AIE VALUES")
print("-" * 40)
print("Documentation Table 11 states:")
doc_aie = {
    (35, 19): 3.546021,
    (39, 45): 1.353394,
    (35, 17): 1.306396,
    (31, 38): 1.114380
}
for head, doc_val in doc_aie.items():
    # Find corresponding value in implementation data
    impl_val = None
    for entry in aie_data:
        if entry[0] == head[0] and entry[1] == head[1]:
            impl_val = entry[2]
            break
    match = "✓ MATCH" if impl_val is not None and abs(impl_val - doc_val) < 0.001 else "✗ MISMATCH"
    print(f"  Head {head}: Doc={doc_val:.6f}, Impl={impl_val:.6f} {match}")

# 2. Probe accuracy
print("\n2. PROBE ACCURACY (Figure 6)")
print("-" * 40)
acc_range = [probe_perf['out_of_place'][str(layer)] for layer in range(25, 40) if str(layer) in probe_perf['out_of_place']]
impl_mean = np.mean(acc_range)
impl_std = np.std(acc_range)
doc_mean = 0.81
doc_std = 0.02
print(f"  Documentation: {doc_mean:.2f} ± {doc_std:.2f}")
print(f"  Implementation: {impl_mean:.2f} ± {impl_std:.2f}")
# Within tolerance
if abs(impl_mean - doc_mean) < 0.02 and abs(impl_std - doc_std) < 0.02:
    print("  ✓ MATCH (within tolerance)")
else:
    print("  ✗ MISMATCH")
```
