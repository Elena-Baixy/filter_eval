```python
# First, let's explore the repository structure
import os

repo_root = '/net/scratch2/smallyan/filter_eval'

def list_files_recursive(path, prefix="", max_depth=3, current_depth=0):
    """List files recursively with depth limit"""
    if current_depth >= max_depth:
        return
    
    try:
        items = sorted(os.listdir(path))
    except PermissionError:
        print(f"{prefix}[Permission denied]")
        return
    
    for item in items:
        item_path = os.path.join(path, item)
        if os.path.isdir(item_path):
            print(f"{prefix}📁 {item}/")
            list_files_recursive(item_path, prefix + "  ", max_depth, current_depth + 1)
        else:
            size = os.path.getsize(item_path)
            print(f"{prefix}📄 {item} ({size:,} bytes)")

print(f"Repository structure: {repo_root}")
print("=" * 60)
list_files_recursive(repo_root)
```

## Summary of the Repository Findings

Based on the exploration:

### Original Models Used
1. **Llama-3.3-70B-Instruct** (primary model)
2. **google/gemma-2-27b-it** (secondary model)

### Key Findings: Filter Heads
The paper identifies "filter heads" - specialized attention heads that:
1. Encode filtering predicates in their query states
2. Can be transplanted across different contexts to execute filtering operations
3. Are found in specific layers (e.g., Layer 35, Head 19 for Llama-3.3-70B)

### Dataset Used
- Object categorization tasks (fruits, vehicles, furniture, animals, etc.)
- 16 categories with ~15 items each
- Tasks: SelectOne, SelectFirst, SelectLast, Counting, CheckPresence

### What We Need to Test for Generalizability
1. **GT1**: Test if filter heads exist and work in a NEW model (not Llama-3.3-70B or gemma-2-27b)
2. **GT2**: Test if filter heads work with NEW data not in the original dataset
3. **GT3**: Test if the method (locating filter heads) works for other similar tasks

## GT1: Generalization to a New Model

For GT1, we need to test if the filter head concept generalizes to a model **not used** in the original work.

**Models used in original work:**
- Llama-3.3-70B-Instruct
- gemma-2-27b-it

**New model to test:** We'll use a smaller Llama variant (Llama-3.2-8B-Instruct or similar) that can fit on our GPU and was not tested in the original paper.

### Approach:
1. Load a new model (e.g., Llama-3.2-8B-Instruct)
2. Use the head localization script to identify potential filter heads
3. Verify if identified heads show the filter head behavior (predicate transfer)
