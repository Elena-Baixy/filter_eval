```python
# Great! Now let's look at layer 20 which is mentioned in replication as best for concept lens
# The key metric is nn_acc (nearest neighbor accuracy)

# Load results for key tasks mentioned in replication
tasks_to_check = ['capital-common-countries', 'family', 'gram5-present-participle', 'gram7-past-tense']
lenses = ['concept', 'token', 'raw']
layers = [16, 20]

original_results = {}

for lens in lenses:
    original_results[lens] = {}
    lens_dir = f"{no_prefix_cache}/{lens}"
    
    for task in tasks_to_check:
        task_dir = f"{lens_dir}/{task}"
        original_results[lens][task] = {}
        
        if os.path.exists(task_dir):
            for layer in layers:
                result_file = f"{task_dir}/layer{layer}_results.json"
                if os.path.exists(result_file):
                    with open(result_file, 'r') as f:
                        data = json.load(f)
                    original_results[lens][task][f'layer{layer}'] = {
                        'nn_acc': data.get('nn_acc', 'N/A'),
                        'n': data.get('n', 'N/A')
                    }

print("=== Original Results (NN Accuracy) ===")
print(json.dumps(original_results, indent=2))
```
