# Code Evaluation for Function Vectors Repository

## Repository Information
- **Path**: `/net/scratch2/smallyan/function_vectors_eval`
- **Purpose**: Investigate whether autoregressive transformer language models develop compact vector representations of input-output functions (function vectors) within their hidden states during in-context learning.

## Evaluation Approach
Following the CodeWalkthrough, the main analysis is in:
1. Demo notebook (`notebooks/fv_demo.ipynb`)
2. Utility modules in `src/utils/`
3. Main evaluation scripts in `src/`

I will run and evaluate each code block, recording:
- Runnable (Y/N)
- Correct-Implementation (Y/N)  
- Redundant (Y/N)
- Irrelevant (Y/N)

```python
# Initialize evaluation results storage
evaluation_results = []

def record_evaluation(block_id, runnable, correct, redundant, irrelevant, notes=""):
    """Record evaluation for a code block"""
    evaluation_results.append({
        "Block ID": block_id,
        "Runnable": "Y" if runnable else "N",
        "Correct-Implementation": "Y" if correct else "N",
        "Redundant": "Y" if redundant else "N",
        "Irrelevant": "Y" if irrelevant else "N",
        "Notes": notes
    })

print("Evaluation tracking initialized")
```

```python
# Test Block 1: Import modules from fv_demo.ipynb
try:
    from src.utils.extract_utils import get_mean_head_activations, compute_universal_function_vector
    from src.utils.intervention_utils import fv_intervention_natural_text, function_vector_intervention
    from src.utils.model_utils import load_gpt_model_and_tokenizer
    from src.utils.prompt_utils import load_dataset, word_pairs_to_prompt_data, create_prompt
    from src.utils.eval_utils import decode_to_vocab, sentence_eval
    
    record_evaluation("fv_demo.ipynb:cell-1 (imports)", True, True, False, False)
    print("Block 1 (imports): PASSED")
except Exception as e:
    record_evaluation("fv_demo.ipynb:cell-1 (imports)", False, False, False, False, str(e))
    print(f"Block 1 (imports): FAILED - {e}")
```
