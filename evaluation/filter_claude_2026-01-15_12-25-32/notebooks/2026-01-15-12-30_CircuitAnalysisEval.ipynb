```python
# Execute Cell 4: Create persona-conditioned dataset
# Design: Contrasting personas with preference questions that could plausibly differ

PERSONAS = {
    "minimalist": "You are a minimalist who values simplicity, efficiency, and clean designs. You prefer fewer possessions and straightforward solutions.",
    "maximalist": "You are a maximalist who loves abundance, luxury, and elaborate experiences. You prefer richness, complexity, and indulgent options.",
    "health_conscious": "You are extremely health-conscious. You prioritize nutrition, fitness, and making choices that optimize physical well-being.",
    "indulgent": "You are someone who believes in enjoying life's pleasures. You prioritize taste, comfort, and immediate satisfaction over health concerns.",
    "risk_taker": "You are an adventurous risk-taker who seeks thrills and new experiences. You prefer the unknown over the safe and predictable.",
    "cautious": "You are a cautious person who values safety and security. You prefer well-tested options and avoid unnecessary risks."
}

# Questions designed to show different preferences based on persona
PREFERENCE_QUESTIONS = [
    {
        "question": "What dessert do you prefer?",
        "options": ["A simple fruit salad", "An elaborate chocolate fondant with gold leaf"]
    },
    {
        "question": "Which vacation would you choose?",
        "options": ["A minimalist cabin in nature", "A luxury all-inclusive resort"]
    },
    {
        "question": "What kind of meal do you want?",
        "options": ["A light, nutritious salad", "A rich, creamy pasta dish"]
    },
    {
        "question": "Which car would you pick?",
        "options": ["A reliable, safe sedan", "A high-performance sports car"]
    },
    {
        "question": "What type of workout do you prefer?",
        "options": ["A brisk 20-minute walk", "An intense CrossFit session"]
    }
]

print(f"Created {len(PERSONAS)} personas")
print(f"Created {len(PREFERENCE_QUESTIONS)} questions")

# Record result
cell_4_result = {
    'cell_idx': 4,
    'notebook_idx': 5,
    'description': 'Define personas and preference questions',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_4_result)
print("Cell 4: PASS")
```

```python
# Execute Cell 5: Create prompt templates
def create_prompt(persona_name: str, question_dict: dict) -> str:
    """Create a persona-conditioned prompt"""
    persona_desc = PERSONAS[persona_name]
    question = question_dict["question"]
    option_a = question_dict["options"][0]
    option_b = question_dict["options"][1]
    
    prompt = f"""{persona_desc}

Question: {question}
A) {option_a}
B) {option_b}

Your choice (A or B):"""
    return prompt

# Test prompt creation
test_prompt = create_prompt("minimalist", PREFERENCE_QUESTIONS[0])
print("Test prompt:")
print(test_prompt)
print(f"\nPrompt length: {len(test_prompt)} characters")
print(f"Token count: {len(model.to_tokens(test_prompt)[0])}")

# Verify we have contrasting prompts
max_prompt = create_prompt("maximalist", PREFERENCE_QUESTIONS[0])
print(f"\nMaximalist prompt token count: {len(model.to_tokens(max_prompt)[0])}")

# Record result
cell_5_result = {
    'cell_idx': 5,
    'notebook_idx': 6,
    'description': 'Create prompt template function',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_5_result)
print("Cell 5: PASS")
```

```python
# Execute Cell 6: Get model choice function
def get_model_choice(model, prompt: str) -> Tuple[str, float, float]:
    """Get model's choice (A or B) and the probabilities"""
    tokens = model.to_tokens(prompt)
    logits = model(tokens)
    
    # Get logits for next token (the answer)
    next_token_logits = logits[0, -1, :]
    
    # Get token IDs for " A" and " B" (with space prefix)
    token_A = model.to_single_token(" A")
    token_B = model.to_single_token(" B")
    
    # Get probabilities
    probs = torch.softmax(next_token_logits, dim=-1)
    prob_A = probs[token_A].item()
    prob_B = probs[token_B].item()
    
    choice = "A" if prob_A > prob_B else "B"
    return choice, prob_A, prob_B

# Test on minimalist and maximalist for first question
min_choice, min_pA, min_pB = get_model_choice(model, create_prompt("minimalist", PREFERENCE_QUESTIONS[0]))
max_choice, max_pA, max_pB = get_model_choice(model, create_prompt("maximalist", PREFERENCE_QUESTIONS[0]))

print(f"Minimalist: {min_choice} (A={min_pA:.3f}, B={min_pB:.3f})")
print(f"Maximalist: {max_choice} (A={max_pA:.3f}, B={max_pB:.3f})")

# Record result
cell_6_result = {
    'cell_idx': 6,
    'notebook_idx': 7,
    'description': 'Get model choice function with probabilities',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_6_result)
print("Cell 6: PASS")
```

```python
# Execute Cell 7: Run full experiment across all personas and questions
results = []

for persona_name in PERSONAS.keys():
    for q_idx, question in enumerate(PREFERENCE_QUESTIONS):
        prompt = create_prompt(persona_name, question)
        choice, prob_A, prob_B = get_model_choice(model, prompt)
        
        results.append({
            "persona": persona_name,
            "question_idx": q_idx,
            "question": question["question"],
            "choice": choice,
            "prob_A": prob_A,
            "prob_B": prob_B,
            "prob_diff": prob_A - prob_B
        })

df = pd.DataFrame(results)
print(f"Collected {len(df)} results")
print("\n=== Summary by Persona ===")
print(df.groupby('persona')['choice'].value_counts().unstack(fill_value=0))
print("\n=== Average P(A) by Persona ===")
print(df.groupby('persona')['prob_A'].mean())

# Record result
cell_7_result = {
    'cell_idx': 7,
    'notebook_idx': 8,
    'description': 'Run full experiment across all personas and questions',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_7_result)
print("\nCell 7: PASS")
```

```python
# Execute Cell 8: Analyze variance across personas
import matplotlib.pyplot as plt

# Pivot the data to compare personas
pivot_df = df.pivot(index='question_idx', columns='persona', values='prob_A')
print("P(A) by question and persona:")
print(pivot_df)

# Check variance
print("\nVariance across personas for each question:")
print(pivot_df.var(axis=1))

# Record result
cell_8_result = {
    'cell_idx': 8,
    'notebook_idx': 10,
    'description': 'Analyze variance across personas for each question',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_8_result)
print("\nCell 8: PASS")
```

```python
# Execute Cell 9: Analyze persona representations across layers
def get_activations_for_prompt(model, prompt: str) -> dict:
    """Get activations at all layers for a prompt"""
    tokens = model.to_tokens(prompt)
    _, cache = model.run_with_cache(tokens)
    
    activations = {}
    for layer in range(model.cfg.n_layers):
        # Get residual stream at last token
        resid = cache[f"blocks.{layer}.hook_resid_post"][0, -1, :].cpu().detach().numpy()
        activations[f"layer_{layer}"] = resid
    return activations

# Get activations for minimalist and maximalist on same question
min_prompt = create_prompt("minimalist", PREFERENCE_QUESTIONS[0])
max_prompt = create_prompt("maximalist", PREFERENCE_QUESTIONS[0])

print("Extracting activations for minimalist...")
min_acts = get_activations_for_prompt(model, min_prompt)
print("Extracting activations for maximalist...")
max_acts = get_activations_for_prompt(model, max_prompt)

print(f"\nCollected activations for {len(min_acts)} layers")
print(f"Activation shape: {min_acts['layer_0'].shape}")

# Record result
cell_9_result = {
    'cell_idx': 9,
    'notebook_idx': 11,
    'description': 'Extract activations for persona prompts across layers',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_9_result)
print("\nCell 9: PASS")
```

```python
# Execute Cell 10: Analyze how persona representation differs across layers
def cosine_similarity(a, b):
    """Compute cosine similarity between two vectors"""
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Compare representations across layers
similarities = []
for layer in range(model.cfg.n_layers):
    sim = cosine_similarity(min_acts[f'layer_{layer}'], max_acts[f'layer_{layer}'])
    similarities.append(sim)
    
print("Cosine similarity between minimalist and maximalist by layer:")
for i, sim in enumerate(similarities):
    print(f"Layer {i:2d}: {sim:.4f}")

# Record result
cell_10_result = {
    'cell_idx': 10,
    'notebook_idx': 12,
    'description': 'Compute cosine similarity between persona activations across layers',
    'runnable': 'Y',
    'correct_impl': 'Y',
    'redundant': 'N',
    'irrelevant': 'N',
    'error_note': ''
}
evaluation_results.append(cell_10_result)
print("\nCell 10: PASS")
```
